````python
import os
import pandas as pd
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import PromptTemplate
from sklearn.metrics import accuracy_score, classification_report

# 1. Charger la clé API depuis le fichier .env
load_dotenv(override=True)

# 2. Initialiser le modèle (temperature=0 pour des réponses déterministes)
llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    temperature=0
)


df = pd.read_csv("IMDB Dataset.csv")


df_sample = df.sample(20, random_state=42).copy()

# 4. Concevoir le prompt de classification
template = """Tu es un classifieur de sentiment binaire spécialisé dans les critiques de films.
Analyse la critique suivante et détermine si elle est positive ou négative.

Règles strictes :
- Réponds UNIQUEMENT par l'un de ces deux mots : "positive" ou "negative".
- N'ajoute aucune ponctuation, majuscule inutile, ni explication.

Critique :
\"\"\"{review}\"\"\"

Sentiment :"""

prompt = PromptTemplate.from_template(template)

# Créer la chaîne LangChain
chain = prompt | llm

# 5. Fonction pour prédire une ligne
def predire_sentiment(texte):
    try:
        # Tronquer le texte si la critique est exceptionnellement longue
        texte_tronque = texte[:2000]
        reponse = chain.invoke({"review": texte_tronque})
        
        # Nettoyage de la réponse pour garantir "positive" ou "negative"
        prediction = reponse.content.strip().lower()
        if "positive" in prediction:
            return "positive"
        elif "negative" in prediction:
            return "negative"
        return prediction
    except Exception as e:
        print(f"Erreur lors de l'appel : {e}")
        return "error"

print("Classification en cours...")
df_sample["prediction"] = df_sample["review"].apply(predire_sentiment)

# 6. Afficher les résultats et métriques
print("\n--- Aperçu des prédictions ---")
print(df_sample[["sentiment", "prediction"]].head(10))

# Calcul des performances
acc = accuracy_score(df_sample["sentiment"], df_sample["prediction"])
print(f"\nExactitude (Accuracy) : {acc * 100:.2f}%")
print("\nRapport détaillé :")
print(classification_report(df_sample["sentiment"], df_sample["prediction"]))
exemple = df_sample.iloc[0]

print("=" * 60)
print("EXEMPLE DE CLASSIFICATION PAR LLM")
print("=" * 60)
print(f"Critique originale :\n\"{exemple['review'][:400]}...\"\n")
print(f"Vrai sentiment attendu : {exemple['sentiment']}")
print(f"Prédiction du LLM      : {exemple['prediction']}")

resultat = "CORRECT" if exemple['sentiment'] == exemple['prediction'] else "INCORRECT"
print(f"Statut                 : {resultat}")
print("=" * 60)
````

````python
import os
import re
import pandas as pd
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import PromptTemplate
from sklearn.metrics import accuracy_score, classification_report

# 1. Configuration de l'environnement et du LLM
load_dotenv(override=True)
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash", temperature=0)

# 2. Définition du dictionnaire d'adjectifs / termes polarisés
LEXIQUE_POSITIF = {
    "good", "great", "excellent", "amazing", "wonderful", "fantastic", 
    "masterpiece", "brilliant", "awesome", "superb", "enjoyable", 
    "beautiful", "perfect", "loved", "entertaining", "charming"
}

LEXIQUE_NEGATIF = {
    "bad", "terrible", "awful", "horrible", "worst", "boring", 
    "poor", "disappointing", "waste", "stupid", "annoying", 
    "ridiculous", "dull", "mess", "unwatchable", "lame"
}

# 3. Fonction de classification heuristique déterministe
def classifier_heuristique(texte):
    # Nettoyage basique : mise en minuscules et extraction des mots
    mots = re.findall(r'\b[a-z]+\b', texte.lower())
    
    score_positif = sum(1 for mot in mots if mot in LEXIQUE_POSITIF)
    score_negatif = sum(1 for mot in mots if mot in LEXIQUE_NEGATIF)
    
    if score_positif > score_negatif:
        return "positive"
    elif score_negatif > score_positif:
        return "negative"
    else:
        # En cas d'égalité ou absence de mots-clés, règle par défaut
        return "negative"

# 4. Fonction de classification par LLM (Prompt)
prompt_template = PromptTemplate.from_template(
    """Classifie cette critique de film comme 'positive' ou 'negative'.
Réponds UNIQUEMENT avec l'un de ces deux mots.

Critique : {review}
Sentiment :"""
)
chaine_llm = prompt_template | llm

def classifier_llm(texte):
    try:
        rep = chaine_llm.invoke({"review": texte[:2000]}).content.strip().lower()
        return "positive" if "positive" in rep else "negative"
    except Exception:
        return "error"

# 5. Application sur un échantillon IMDb
df = pd.read_csv("IMDB Dataset.csv")
df_test = df.sample(30, random_state=42).copy()

print("1/2 - Calcul des prédictions heuristiques...")
df_test["pred_heuristique"] = df_test["review"].apply(classifier_heuristique)

print("2/2 - Calcul des prédictions LLM...")
df_test["pred_llm"] = df_test["review"].apply(classifier_llm)

# 6. Comparaison des résultats
acc_heuristique = accuracy_score(df_test["sentiment"], df_test["pred_heuristique"])
acc_llm = accuracy_score(df_test["sentiment"], df_test["pred_llm"])

print("\n================ RÉSULTATS COMPARATIFS ================")
print(f"Exactitude Heuristique (Dictionnaire) : {acc_heuristique * 100:.2f}%")
print(f"Exactitude LLM (Gemini Prompt)         : {acc_llm * 100:.2f}%")
print("=======================================================\n")

# Affichage des cas où l'heuristique s'est trompée mais pas le LLM
erreurs_heuristique = df_test[
    (df_test["pred_heuristique"] != df_test["sentiment"]) & 
    (df_test["pred_llm"] == df_test["sentiment"])
]
print(f"Nombre de cas complexes résolus par le LLM mais manqués par l'heuristique : {len(erreurs_heuristique)}")
````
